## List:
A list is a collection used to store multiple items in a single variable.

Lists are:

Ordered
Mutable (changeable)
Allow duplicate values
Can store different data types

```python
numbers = [1, 2, 3]
names = ["Alice", "Bob"]
mixed = [1, "Hello", 3.5, True]
empty = []
```

```python
print(numbers)
print(names)
print(mixed)
print(empty)
```

```python
## Accessing list elements
fruits = ["apple", "banana", "mango"]
print(fruits[0])   
print(fruits[1])
print(fruits[-1])   
```

```python
## Slicing lists
print(fruits[0:2])
print(fruits[:2])
print(fruits[1:])
```

```python
## Changing list elements
colors = ["red", "blue", "green"]

colors[1] = "yellow"

print(colors)
```

```python
## Adding elements to a list
colors.append("purple")
print(colors)
```

```python
## Add an element at a specific position
colors.insert(1, "orange")
print(colors)
```

```python
## Add multiple elements to a list
colors.extend(["pink", "cyan"])
print(colors)
```

```python
## Removing elements from a list
colors.remove("yellow")
print(colors)
```

```python
## Removing an element by index
del colors[2]
print(colors)
```

```python
## remove that element and return it
removed_color = colors.pop(1)
print("Removed color:", removed_color)
```

```python
## remove last element and return it
remove_color = colors.pop()
print("Removed color:", remove_color)
```

```python
col = fruits.clear()
print(col)
```

```python
## looping through a list
animals = ["cat", "dog", "lion"]

for animal in animals:
    print(animal)\
        
## using index
for i in range(len(animals)):
    print(i, animals[i])
```

```python
lst=[]
for i in range(10):
    lst.append(i**2)
    
print(lst)
```

```python
## shortcut for creating a list using list comprehension
## syntax: [expression for item in iterable]
lst=[i**2 for i in range(10)]
print(lst)
```

```python
## with condition
even = [x for x in range(10) if x % 2 == 0]
print(even)
```

```python
##enumerate() 
##enumerate() is used when you need both the index and the value while looping through a sequence (list, tuple, string, etc.).
fruits = ["apple", "banana", "mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
```

```python
## List length
numbers = [1, 2, 3, 4]

print(len(numbers))
```

```python
## checking item existence in a list
fruits = ["apple", "banana"]

print("apple" in fruits)
```

```python
## sorting a list
nums = [4, 1, 3, 2]

nums.sort()

print(nums)
```

```python
## descending order
print(nums.sort(reverse=True))
```

```python
## reverse a list
nums.reverse()
print(nums)
```

```python
## copying a list
a = [1, 2, 3]
b = a.copy()
```

```python
## Nested lists
matrix = [[1, 2],[3, 4]]
print(matrix[0][1])
```

```python
## useful functions
nums = [1, 2, 3, 4]

print(max(nums))
print(min(nums))
print(sum(nums))
```

```python
## Nested list comprehension
pairs = [(i, j) for i in range(3) for j in range(2)]

print(pairs)
```
