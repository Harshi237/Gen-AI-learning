VARIABLES:

A variable in Python is a name used to store data or values in memory. It acts like a container that holds information, and the value can change during program execution.

```python
a=100
```

```python
## Declaring And Assigning Variables
age=18
height=5.7
name="Anushka"
is_student=True

## printing the variables
print("age :",age)
print("Height:",height)
print("Name:",name)
```

```python
## Naming Conventions
## Variable names should be descriptive
## They must start with a letter or an '_' 
## It contains letter,numbers and underscores
## variables names case sensitive

#valid variable names

first_name="Harshita"
last_name="Gupta"
```

```python
# # Invalid variable names
# 2age=18
# first-name="Harshita"
# @name="Harshi"
```

```python
## case sensitivity
name="Ram"
Name="Ravi"

print(name==Name)
```

```python
## Understnading Variable types
## Python is dynamically typed
# and type of a variable is determined at runtime
age=25 #int
height=5.6 #float
name="Manas" #str
is_student=True #bool

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```
