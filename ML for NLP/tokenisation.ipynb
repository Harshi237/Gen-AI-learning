## Some Basic Termonologies

1. Corpus: Paragraph
2. Documents: Sentences
3. Vocabulary: Unique Word that is present in the dictionary
4. Words: All the words that are present in the corpus

## Tokenization

Tokenisation (or tokenization) is the process of breaking a sentence, paragraph, or program code into smaller units called tokens.

Paragraph -> Sentence
Sentence -> Words

example::

Paragraph: I like Mango. My friend like Apple.
    |
    | tokenisation
    |
    v
Sentences: 1.I like Mango
           2.My Friend like Apple 

here total words are 7
but total vocabulary are 6 (unique words: here like appear 2 times ..but we count only 1 time)

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
corpus = """ I Like Mangoes.
My fried like Apples. """
```

```python
print(corpus)
```

```python
## Tokenization
## Paragraph-> Sentences

from nltk.tokenize import sent_tokenize
```

```python
documents = sent_tokenize(corpus)
```

```python
for sentence in documents:
    print(sentence)
```

```python
## Tokenization
## paragraph --> words
## sentence --> words

from nltk.tokenize import word_tokenize
```

```python
words = word_tokenize(corpus)
```

```python
print(words)
```
